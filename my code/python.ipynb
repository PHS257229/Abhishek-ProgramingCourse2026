break and continue 

```python
for i in range(1, 20):
    if i == 6:
        break 
    print(i)
print("stopped")
```

```python
for i in range(1,20):
    if i==6:
        continue
    print(i)
```

nested loops 


```python
for row in range(2):
    for col in range(3):
        print(row, col)

for i in range(1,5):
    print("*"*i)
```

# Lists : python's Arrays 

```python
r = [9.8, 9.85, 9.79,9.83]
print(r[0])
print(r[-1])
print(len(r))
r.append(9.9)
print(r[:])

```

```python
grid= [[1.2, 1.3,1.8],
       [2.1,2.4,2.7]]
print(grid[1][2])

for r in range(2):
    for c in range(3):
        print(grid[r][c])

for i in range(5):
    print(i)
```

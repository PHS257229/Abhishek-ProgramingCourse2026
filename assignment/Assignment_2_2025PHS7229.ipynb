**Question 2.** A ball is thrown straight up at u = 20 m/s. Its height is h = u t − ½ g t². Rearranged, (g/2) t² − u t + h = 0.

Test with h = 15 m and h = 25 m.

```python
# Question 2
import math

u = 20
g = 9.81

for h in [15, 25]:

    D = u**2 - 2*g*h

    print(f"\nHeight = {h} m")

    if D > 0:
        t1 = (u - math.sqrt(D)) / g
        t2 = (u + math.sqrt(D)) / g

        print("Ball reaches the height twice")
        print(f"First time = {t1:.2f} seconds")
        print(f"Second time = {t2:.2f} seconds")

    elif D == 0:
        t = u / g
        print("Ball just touches the height at the top")
        print(f"Time = {t:.2f} seconds")

    else:
        max_height = u**2 / (2*g)
        print("Ball never reaches this height")
        print(f"Maximum height = {max_height:.2f} m")
```

```python
# Question 10
import math

def proj_range(v0, theta_deg=45.0, g=9.81):
    theta_rad = math.radians(theta_deg)

    R = (v0**2 * math.sin(2 * theta_rad)) / g

    return R


# Earth at 45 degrees
r1 = proj_range(20, 45)

# Earth at 30 degrees
r2 = proj_range(20, 30)

# Moon at 45 degrees
r3 = proj_range(20, 45, 1.62)

# Mars at 60 degrees
r4 = proj_range(20, 60, 3.71)


print(f"Range on Earth at 45 degrees = {r1:.2f} m")
print(f"Range on Earth at 30 degrees = {r2:.2f} m")
print(f"Range on Moon at 45 degrees = {r3:.2f} m")
print(f"Range on Mars at 60 degrees = {r4:.2f} m")
```

**Question 13.** Write factorial(n) using a for loop. Then write exp_series(x, nterms=10) that adds the first nterms terms of Σ xⁿ / n!, calling factorial inside it.

For x = 2, print the approximation and its error compared with math.exp(2) for nterms = 2, 4, 8 and 16.

```python
# Question 13
import math


def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


def exp_series(x, nterms=10):
    total = 0

    for n in range(nterms):
        term = (x**n) / factorial(n)
        total = total + term

    return total


# Calculate for x = 2
x = 2

# Actual value of e^2
actual_value = math.exp(x)

# Test different numbers of terms
for nterms in [2, 4, 8, 16]:
    approximate_value = exp_series(x, nterms)
    error = abs(actual_value - approximate_value)

    print(f"Terms = {nterms}")
    print(f"Approximate value = {approximate_value:.10f}")
    print(f"Actual value = {actual_value:.10f}")
    print(f"Error = {error:.10f}")
    print()
```

**Question 14.** Newton's method for √a repeats  x → ½ (x + a/x)  starting from x = a.

Write newton_sqrt(a, tol=1e-10) that repeats this while |x² − a| > tol and returns both the estimate and the number of iterations.

Test with a = 2, 9 and 10⁶, and then with a = 2 and tol=1e-3.

```python
# Question 14
def newton_sqrt(a, tol=1e-10):
    x = a
    iterations = 0

    while abs(x**2 - a) > tol:
        x = 0.5 * (x + a / x)
        iterations = iterations + 1

    return x, iterations


# Test 1: Square root of 2
result, count = newton_sqrt(2)
print(f"Square root of 2 = {result:.10f}")
print(f"Iterations = {count}")

# Test 2: Square root of 9
result, count = newton_sqrt(9)
print(f"\nSquare root of 9 = {result:.10f}")
print(f"Iterations = {count}")

# Test 3: Square root of 1000000
result, count = newton_sqrt(10**6)
print(f"\nSquare root of 1000000 = {result:.10f}")
print(f"Iterations = {count}")

# Test 4: Square root of 2 with larger tolerance
result, count = newton_sqrt(2, tol=1e-3)
print(f"\nSquare root of 2 (tol=1e-3) = {result:.10f}")
print(f"Iterations = {count}")
```

**Question 15.** Write celsius_to_kelvin(T_C). Then write convert_all(temps) that builds and returns a new list of Kelvin values by calling celsius_to_kelvin on each element.

```python
# Question 15

def celsius_to_kelvin(c):
    return c + 273.15



def convert_all(temperatures):
    kelvin_values = []

    for c in temperatures:
        k = celsius_to_kelvin(c)
        kelvin_values.append(k)

    return kelvin_values



temperatures = [-196, 0, 25, 100]

kelvin_values = convert_all(temperatures)

print(f"{'Celsius (°C)':<15} {'Kelvin (K)':<15}")
print("-" * 30)

for c, k in zip(temperatures, kelvin_values):
    print(f"{c:<15.2f} {k:<15.2f}")
```

**Question 20.** The trapezoidal rule approximates ∫ y dx by Σ dx (yᵢ + yᵢ₊₁)/2. With arrays, y[:-1] holds every left point and y[1:] every right point.

(a) Use 101 points to integrate sin x from 0 to π (exact answer 2).

(b) Find the work done by a spring force F = kx with k = 200 N/m from x = 0 to 0.1 m, and compare with ½ k x².

```python
# Question 20
import numpy as np

# Part (a): Integrate sin(x) from 0 to pi

x = np.linspace(0, np.pi, 101)
y = np.sin(x)

dx = x[1] - x[0]

# Apply trapezoidal rule
integral = np.sum(dx * (y[:-1] + y[1:]) / 2)

# Exact answer
exact_integral = 2.0

print("Part (a): Integration of sin(x)")
print(f"Numerical integral = {integral:.8f}")
print(f"Exact integral = {exact_integral:.8f}")
print(f"Error = {abs(exact_integral - integral):.8f}")


# Part (b): Work done by a spring

k = 200  # Spring constant in N/m

# Create position array from 0 to 0.1 m
x_spring = np.linspace(0, 0.1, 101)

# Calculate spring force F = k*x
F = k * x_spring

dx_spring = x_spring[1] - x_spring[0]

# Calculate work using trapezoidal rule
work = np.sum(dx_spring * (F[:-1] + F[1:]) / 2)

# Exact work = 1/2 * k * x^2
exact_work = 0.5 * k * (0.1)**2

print("\nPart (b): Work done by spring")
print(f"Numerical work = {work:.8f} J")
print(f"Exact work = {exact_work:.8f} J")
print(f"Difference = {abs(exact_work - work):.8f} J")
```

**Question 29.** V = [1, 2, 3, 4, 5] volts and I = [0.21, 0.39, 0.62, 0.80, 1.01] amperes.

Write them to ohm.csv using open(), a for loop and f.write(): put the header line V,I first, then one line per measurement.

Read the file back with np.loadtxt and compute R = V/I for every row, using the columns of the loaded array. Print each R and the mean R.

```python
# Question 29
import numpy as np


V = np.array([1, 2, 3, 4, 5])
I = np.array([0.21, 0.39, 0.62, 0.80, 1.01])


with open("ohm.csv", "w") as file:
    file.write("V,I\n")

    for i in range(len(V)):
        file.write(f"{V[i]},{I[i]}\n")

data = np.loadtxt("ohm.csv", delimiter=",", skiprows=1)

voltage = data[:, 0]
current = data[:, 1]


R = voltage / current

# Print individual resistance values
for i in range(len(R)):
    print(f"Resistance {i+1} = {R[i]:.2f} ohms")

# Calculate and print mean resistance
mean_R = np.mean(R)

print(f"Mean resistance = {mean_R:.2f} ohms")
```

**Question 30.** Make t = np.linspace(0, 2, 201) and x = cos(2πt).

Save x with np.save, load it back, and print the shape and the largest difference from the original.

Save t and x together with np.savez("run1.npz", time=t, position=x). Load the file and print the first three times and the last position.

```python
# Question 30
import numpy as np

t = np.linspace(0, 2, 201)

x = np.cos(2 * np.pi * t)

# Save x array in a .npy file
np.save("run1.npy", x)

x_loaded = np.load("run1.npy")

print("Shape:", x_loaded.shape)
print("Largest difference:", np.max(np.abs(x - x_loaded)))

# Save both arrays in a .npz file
np.savez("run1.npz", time=t, position=x)

data = np.load("run1.npz")

# Print requested values
print("First three time values:", data["time"][:3])
print("Last position value:", data["position"][-1])
```

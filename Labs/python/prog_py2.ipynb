## 1. if / elif / else

```python
score = 72

if score >= 90:
    print("A")
elif score >= 60:
    print("Pass")
else:
    print("Fail")
```

## 2. Writing good conditions

Try changing `x` and `name`.

```python
x = 15
if 10 <= x <= 20:          # chained comparison
    print("in range")

if "a" in "banana":        # membership
    print("found")

name = ""
if not name:               # empty string is falsy
    print("empty")

print(bool(0), bool(""), bool(None), bool([]), bool(5))
```

## 3. Nested if and logical operators

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Show your ID")
else:
    print("Too young")

if age >= 18 and has_id:
    print("Same check in one line")
```

## 4. while loop

Python has no `++`: use `+= 1`.

```python
reading = 1
while reading <= 5:
    print("Taking reading #", reading)
    reading += 1

print("Data collection complete.")
```

## 5. for loop with range()

```python
for reading in range(6,0,-1):
    print("Taking reading #", reading)

print("Data collection complete.")
```

## 6. range() variations

The stop value is never included.

```python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 10, 2)))
print(list(range(5, 0, -1)))
```

## 7. break

```python
for i in range(1, 11):
    if i == 6:
        break          # exit the loop right here
    print("Reading #", i)

print("Stopped early.")
```

## 8. continue

```python
for i in range(1, 6):
    if i == 3:
        continue       # skip the rest of THIS round only
    print("Reading #", i)
```

## 9. Nested loops

```python
for row in range(2):
    for col in range(3):
        print("row", row, "col", col)

for i in range(1, 5):
    print("*" * i)
```

## 10. Lists

Lists replace C++ arrays and can grow.

```python
readings = [9.8, 9.85, 9.79, 9.83, 9.81]

print(readings[0])        # first
print(readings[-1])       # last
print(len(readings))      # size
print(readings[1:3])      # slice: items 1 and 2

readings.append(9.9)
readings[0] = 10.0        # lists are changeable
print(readings)
```

## 11. Looping over a list

```python
readings = [9.8, 9.85, 9.79, 9.83, 9.81]

for r in readings:
    print(r)

for i in range(len(readings)):
    print(i, readings[i])
```

## 12. List methods and functions

```python
nums = [5, 2, 9, 1]

nums.append(7)
nums.insert(0, 100)
nums.remove(9)
last = nums.pop()
print(nums, "popped:", last)

nums.sort()
print(nums)
print(sum(nums), min(nums), max(nums))
```

## 13. 2D lists (grids)

Same idea as `detectorGrid` in the C++ class.

```python
grid = [[1.2, 1.5, 1.8],
        [2.1, 2.4, 2.7]]

print(grid[1][2])

for row in range(2):
    for col in range(3):
        print(f"grid[{row}][{col}] = {grid[row][col]}")
```

## 16. The random module

Run the cell several times: the output changes each time.

```python
import random

print(random.random())             # 0 to 1
print(random.randint(1, 6))        # dice roll
print(random.choice(["a", "b", "c"]))
```

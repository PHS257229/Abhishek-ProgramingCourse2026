# Maths functions: use np, not math


1. With x = np.linspace(0, np.pi, 3) , print np.sin(x) , np.exp(np.array([0.0,
1.0])) and np.sqrt(np.array([4.0, 9.0])) .
2. For 10 values of x between 0 and 2 pi, compute np.sin(x)**2 + np.cos(x)**2 and
print the result.
3. Print np.log of the array [1, e, e squared] . Hint: np.e .
4. Take the angles 0, 30, 45, 60, 90 degrees, convert them with np.radians and print
their sines

```python

import numpy as np

# Question 1: sin, exp and sqrt
x = np.linspace(0, np.pi, 3)

print("Sin values:", np.sin(x))
print("Exponential values:", np.exp(np.array([0.0, 1.0])))
print("Square root values:", np.sqrt(np.array([4.0, 9.0])))


# Question 2: sin(x)^2 + cos(x)^2
x = np.linspace(0, 2 * np.pi, 10)

result = np.sin(x)**2 + np.cos(x)**2

print("\nSin^2(x) + Cos^2(x):")
print(result)


# Question 3: Natural logarithm of [1, e, e^2]
a = np.array([1, np.e, np.e**2])

print("\nLogarithm values:", np.log(a))


# Question 4: Sine of angles in degrees
angles = np.array([0, 30, 45, 60, 90])

radians = np.radians(angles)

print("\nAngles in degrees:", angles)
print("Sine values:", np.sin(radians))
```

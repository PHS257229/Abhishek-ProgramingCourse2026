Question 5 : Ask for a number and count its digits using //

```python
num = int(input("Enter a number: "))
count = 0

if num == 0:
    count = 1
else:
    temp = abs(num) # Handles negative numbers by converting them to positive
    while temp > 0:
        temp //= 10 # Integer division removes the last digit
        count += 1  # Increments digit counter

print("Number of digits:", count)
```

sum the degits of a number


```python
# Take input from the user
num = int(input("Enter a number: "))

# Handle negative numbers
num = abs(num)

digit_sum = 0

# Loop until all digits are processed
while num > 0:
    digit = num % 10       # Extract the last digit
    digit_sum += digit     # Add the digit to sum
    num = num // 10        # Remove the last digit

print("Sum of digits:", digit_sum)
```

Question 6: Reverse a number (1234 becomes 4321) using % and //.

```python
num = int(input("Enter a number to reverse: "))
reversed_num = 0
temp = num

while temp > 0:
    digit = temp % 10              # Step 1: Extract the last digit
    reversed_num = reversed_num * 10 + digit  # Step 2: Build the reversed number
    temp //= 10                    # Step 3: Remove the last digit from temp

print("Reversed number:", reversed_num)
```

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

Question 3: Ask for n and print its factorial.

```python
n = int(input("Enter the number for factorial :  "))
factorial = 1

for i in range(1 , n+1):
    factorial *= i

print("factorial of number ",n,"is : ", factorial)
```

Question 1 : Print this pattern: 1, 12, 123, 1234, each on its own line.

```python
for i in range(1, 5):
    for j in range(1, i + 1):
        print(j, end="")
    print()  # Moves to the next line after each row
```

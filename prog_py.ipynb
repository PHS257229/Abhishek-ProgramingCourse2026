## 2. Variables and types

```python
name = "Asha"
age = 20
height = 1.65
is_student = True

print(name, type(name))
print(age, type(age))
print(height, type(height))
print(is_student, type(is_student))
```

## 3. Changing a variable's type

```python
x = 5
print(x, type(x))

x = "five"          # same variable, new type (dynamic typing)
print(x, type(x))
```

## 4. Type conversion

```python
a = "10"
b = int(a) + 5      # str -> int
print(b)

c = float(b)        # int -> float
print(c)
```

## 5. Input

Type your answers in the box that appears.

```python
name = input("Name: ")
age = int(input("Age: "))    # input() returns str, so convert

print("Hello", name)
print(f"{name} is {age} years old")
```

## 6. Arithmetic operators

```python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)        # always float
print(7 // 2)       # floor division
print(7 % 2)        # remainder
print(7 ** 2)       # power
```

## 7. Comparison and logical operators

```python
x = 7
print(x == 7)
print(x != 7)
print(x > 3 and x < 10)
print(x < 3 or x > 5)
print(not x > 5)
```
